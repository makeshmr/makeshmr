Binary Tree representation in python




```
# This is formatted as code


class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

# Example usage:
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)

```
