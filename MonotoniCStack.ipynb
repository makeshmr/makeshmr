```python
def nge(arr):

  stack = []
  res = [0]*len(arr)

  #traversing reverse order --> len(nums)-1 --> 0

  for i in range(len(arr)-1,-1,-1):

    while stack and stack[-1] <= arr[i]:
      stack.pop()
    if len(stack) == 0:
      res[i]=-1
    else:
      stack[-1]>arr[i]
      res[i]=stack[-1]

    stack.append(arr[i])


  return res
```

```python
#using stack

def previous_smallerST(arr):

  stack = []
  res  = [0]*len(arr)

  for i in range(len(arr)):

    while stack and stack[-1]>arr[i]:
      stack.pop()

    if not stack:
      res[i] = -1

    if stack and stack[-1]<arr[i]:
      res[i] = stack[-1]

    stack.append(arr[i])

  return res
```
