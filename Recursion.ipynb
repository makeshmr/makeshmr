**Q**. From N to 1 you need to print all the numbers via recursion?


````python
def printNums(N):
  if N < 1:
    return

  print(N,end=" ")
  printNums(N-1)
````

````python
N = 3
printNums(3)
````

**Call Stack (Top to Bottom):**

| Functions     |   Description                  |
| ------------- | -----------------------------  |
| printNums(0)  |   Base case (returns)          |
| printNums(1)  |   Prints 1, calls printNums(0) |
| printNums(2)  |   Prints 2, calls printNums(1) |
| printNums(3)  |   Prints 3, calls printNums(2) |


Initial Call: printNums(3)
- Checks: N < 1 (False)
- Prints: 3
- Calls: printNums(2)

Second Call: printNums(2)
- Checks: N < 1 (False)
- Prints: 2
- Calls: printNums(1)

Third Call: printNums(1)
- Checks: N < 1 (False)
- Prints: 1
- Calls: printNums(0)

Fourth Call: printNums(0)
- Checks: N < 1 (True! Base case reached)

Returns: Back to printNums(1)

- Returning from Calls
- printNums(1) returns to printNums(2)
- printNums(2) returns to printNums(3)
- printNums(3) returns to the original caller


````python
def factorial(n):

  if n <= 1:
    return 1

  return n*factorial(n-1)
````

````python
factorial(5)
````

- When factorial(5) is called, it goes through multiple recursive calls until it reaches the base case (n <= 1).

- Each function call is pushed onto the call stack, and once the base case is reached, the calls start resolving (popping off the stack).

Call Stack Representation
Recursive Calls (Pushing onto the Stack)

1.      factorial(5) → Calls factorial(4)
2.      factorial(4) → Calls factorial(3)
3.      factorial(3) → Calls factorial(2)
4.      factorial(2) → Calls factorial(1)
5.      factorial(1) → Base case reached, returns 1

At this point, the stack looks like this (last in, first out - LIFO):











At each step, the stack grows and shrinks as follows:
Before Reaching Base Case (Pushing Calls)

```
factorial(5)   
factorial(4)   
factorial(3)   
factorial(2)   
factorial(1)   ⬅ Base case returns 1
```


After Resolving (Popping Calls)


```
factorial(2)   ⬅ Returns 2
factorial(3)   ⬅ Returns 6
factorial(4)   ⬅ Returns 24
factorial(5)   ⬅ Returns 120

```





- The function calls itself recursively until it reaches factorial(1), which stops recursion.

- The stack stores each recursive call until the base case is reached.

- Once the base case returns 1, each function call resolves by multiplying the return values.

- Stack follows LIFO (Last In, First Out).
