**Q**. From N to 1 you need to print all the numbers via recursion?


````python
def printNums(N):
  if N < 1:
    return

  print(N,end=" ")
  printNums(N-1)
````

````python
N = 3
printNums(3)
````

**Call Stack (Top to Bottom):**

| Functions     |   Description                  |
| ------------- | -----------------------------  |
| printNums(0)  |   Base case (returns)          |
| printNums(1)  |   Prints 1, calls printNums(0) |
| printNums(2)  |   Prints 2, calls printNums(1) |
| printNums(3)  |   Prints 3, calls printNums(2) |


Initial Call: printNums(3)
- Checks: N < 1 (False)
- Prints: 3
- Calls: printNums(2)

Second Call: printNums(2)
- Checks: N < 1 (False)
- Prints: 2
- Calls: printNums(1)

Third Call: printNums(1)
- Checks: N < 1 (False)
- Prints: 1
- Calls: printNums(0)

Fourth Call: printNums(0)
- Checks: N < 1 (True! Base case reached)

Returns: Back to printNums(1)

- Returning from Calls
- printNums(1) returns to printNums(2)
- printNums(2) returns to printNums(3)
- printNums(3) returns to the original caller


````python
def factorial(n):

  if n <= 1:
    return 1

  return n*factorial(n-1)
````

````python
factorial(5)
````

- When factorial(5) is called, it goes through multiple recursive calls until it reaches the base case (n <= 1).

- Each function call is pushed onto the call stack, and once the base case is reached, the calls start resolving (popping off the stack).

Call Stack Representation
Recursive Calls (Pushing onto the Stack)

1.      factorial(5) → Calls factorial(4)
2.      factorial(4) → Calls factorial(3)
3.      factorial(3) → Calls factorial(2)
4.      factorial(2) → Calls factorial(1)
5.      factorial(1) → Base case reached, returns 1

At this point, the stack looks like this (last in, first out - LIFO):











At each step, the stack grows and shrinks as follows:
Before Reaching Base Case (Pushing Calls)

```
factorial(5)   
factorial(4)   
factorial(3)   
factorial(2)   
factorial(1)   ⬅ Base case returns 1
```


After Resolving (Popping Calls)


```
factorial(2)   ⬅ Returns 2
factorial(3)   ⬅ Returns 6
factorial(4)   ⬅ Returns 24
factorial(5)   ⬅ Returns 120

```





- The function calls itself recursively until it reaches factorial(1), which stops recursion.

- The stack stores each recursive call until the base case is reached.

- Once the base case returns 1, each function call resolves by multiplying the return values.

- Stack follows LIFO (Last In, First Out).

Sum of N with Recursion



````python
def sum_N(n):
  #base case
  if n <= 1: return 1

  return n + sum_N(n-1)
````

**Q-509. Fibonacci Number**

````python
def fib(n):

  if n == 1 : return 1
  if n == 0 : return 0

  return fib(n-2) + fib(n-1)
````

````python
fib(4)
````

Checking an array is sorted using recursion



````python
def isSorted(arr,n):

  if n == 0 : return True
  if n == 1 : return True

  return arr[n-1] >= arr[n-2] and isSorted(arr,n-1)
````

````python
isSorted([1,2,5,4],len([1,2,5,4]))
````

🔍 How It Works

- It checks whether the last two elements are in order: arr[n-1] >= arr[n-2]
- If true, it recursively checks the rest of the array.
- Base case: If array has 0 or 1 element → it's trivially sorted.

✅ Example: arr = [1, 2, 3, 5, 6], n = 5

### ✅ Call Stack for `isSorted([1, 2, 3, 5, 6], 5)`

| Step | Function Call                 | Condition Checked             | Result       |
|------|-------------------------------|--------------------------------|--------------|
| 1    | isSorted([1,2,3,5,6], 5)      | 6 >= 5                         | True         |
| 2    | isSorted([1,2,3,5], 4)        | 5 >= 3                         | True         |
| 3    | isSorted([1,2,3], 3)          | 3 >= 2                         | True         |
| 4    | isSorted([1,2], 2)            | 2 >= 1                         | True         |
| 5    | isSorted([1], 1)              | Base case → return True        | ✅ Final True |


🧠 Time & Space Complexity
- Time Complexity: O(n)
- Space Complexity: O(n) (due to recursion stack)

🔍 What’s shrinking?
- You're not actually modifying the array, but by decreasing the value of n, you're only looking at a smaller portion of it each time.
- So it feels like the array is shrinking from the end toward the beginning.



344. Reverse String



````python
def reverse_string(s):
  if len(s) == 0: return ''
  if len(s) == 1: return s[0]

  return reverse_string(s[1:]) + s[0]
````


🔍 What’s Happening?
- The function keeps removing the first character **(s[0])** and recursively calling itself on the rest of the string **(s[1:])**.
- Once it reaches the base case, it builds the reversed string by appending characters back in reverse order.

🧩 Example: reverse_string("abcd")
📤 Recursive Calls (Shrinking the input)


### 📊 Recursive Call Stack for reverse_string("abcd")

| Step | Input String | Expression                     | Result So Far |
|------|---------------|---------------------------------|---------------|
| 1    | "abcd"        | reverse_string("bcd") + "a"     |               |
| 2    | "bcd"         | reverse_string("cd") + "b"      |               |
| 3    | "cd"          | reverse_string("d") + "c"       |               |
| 4    | "d"           | Base case → return "d"          | "d"           |
| 5    | return        | "d" + "c"                        | "dc"          |
| 6    | return        | "dc" + "b"                       | "dcb"         |
| 7    | return        | "dcb" + "a"                      | "dcba"        |


📊 Final Complexity Summary

- Time Complexity	O(n²) (due to slicing and concatenation in each recursive call)
- Space Complexity	O(n²) (due to new string creation at each step)

Each recursive call stores a new string of size O(n), leading to O(n²) space if considering all copies.
The recursive call stack uses O(n) space due to n function calls.
However, if we consider immutable string storage overhead, the total space complexity remains O(n²) due to slicing.

Binary search using recursion



````python
class Solution:
    def search(self, nums: List[int], target: int) -> int:
        low = 0
        high = len(nums)-1

        return self.helper(nums,target,low,high)


    def helper(self,nums,target,low,high):

        if low > high:  # Add base case: target not found
            return -1

        mid = low + (high-low)//2

        if nums[mid]== target: return mid

        elif nums[mid]<target:
            low = mid+1
            return self.helper(nums,target,low,high)

        else:
            high = mid -1
            return self.helper(nums,target,low,high)
````

Print all Subsets

````python
class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        idx = 0
        ans = []
        res = []
        n = len(nums)

        self.helper(nums,ans,n,idx,res)

        return res





    def helper(self, nums,ans,n,idx,res):

        #base case
        if (idx == n):
            res.append(ans[:])

            return

        ans.append(nums[idx])
        self.helper(nums,ans,n,idx+1,res)

        ans.pop()
        self.helper(nums,ans,n,idx+1,res)
````

the ans[:] is used to create a copy of the ans list before appending it to res.

Why is ans[:] needed?
In Python, lists are mutable, meaning that if you store a reference to ans in res, any changes made to ans later will also affect the stored value in res.

By using ans[:], a new copy of ans is created, preserving the current state at that moment.
